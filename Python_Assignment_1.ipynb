```python
#String Methods
strM = "Python beginner tutorial"

#uppercase
print(strM.upper())

#lowercase
print(strM.lower())

#Capitalize
print(strM.capitalize())

#number of occurrences of character
print(strM.count("t"))

#Replace
strM = 'Python beginner tutorial'
print(strM.replace("Python", "Machine Learning"))

```

```python
#Tuples (Creation, Modification and Access)
tuple1 = (10, 20, 30)
tuple2 = (40, 50, 60)

print(tuple1)
print(tuple2)

#Concatenate
t_combine = tuple1 + tuple2

print(t_combine)

#Repeat
print(t_combine * 3)

#Access the 3rd element
print(t_combine[2])

#Access the first three elements
print(t_combine[:3])

#Access the last three elements
print(t_combine[-3:])
```
